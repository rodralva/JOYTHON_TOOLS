```python
%matplotlib  widget

gains=[]
ovs=[7,9,12]
for OV in  ovs:
    gains.append(GetGains("DAPHNE_VUV",4,OV,MODE="GAIN"))

# peform a linear fit to the data
x = np.array(ovs)
y = np.array(gains)
m, b = np.polyfit(x, y, 1)

plt.plot(ovs,gains,"s",label="OV=7V") 

x=np.linspace(0,13,100)
plt.plot(x, m*x + b,label="Fit: y={:.2f}x+{:.2f}".format(m,b))

plt.xlim(0,13)
plt.ylim(0,1e7)
plt.grid()


plt.figure()
gains=[]
ovs=[2,2.5,3]
for OV in  ovs:
    gains.append(GetGains("APSAIA_VUV_2",4,OV,MODE="GAIN"))

# peform a linear fit to the data
x = np.array(ovs)
y = np.array(gains)
m, b = np.polyfit(x, y, 1)

plt.plot(ovs,gains,"s",label="OV=7V") 

x=np.linspace(0,13,100)
plt.plot(x, m*x + b,label="Fit: y={:.2f}x+{:.2f}".format(m,b))

plt.xlim(0,5)
plt.ylim(0,1e7)
plt.grid()
```
